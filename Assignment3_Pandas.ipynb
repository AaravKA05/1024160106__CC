## Q6. Create a sample dataset (employees.csv) containing information about employees in a company

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv("employees.csv", index=False)
emp_df
```

### e) Calculate statistics from the dataset

```python
avg_salary = emp_df["Salary"].mean()
total_bonus = emp_df["Bonus"].sum()
youngest_age = emp_df["Age"].min()
highest_rating = emp_df["Rating"].max()

print(f"i.   Average salary of employees   : {avg_salary}")
print(f"ii.  Total bonus paid to employees : {total_bonus}")
print(f"iii. Youngest employee's age       : {youngest_age}")
print(f"iv.  Highest performance rating    : {highest_rating}")
```

### f) Sort the DataFrame by the Salary column in descending order

```python
emp_df_sorted = emp_df.sort_values(by="Salary", ascending=False)
emp_df_sorted
```

### g) Add a new column that categorizes employees based on their performance rating

```python
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance_Category"] = emp_df["Rating"].apply(categorize)
emp_df
```

### h) Identify missing values in the DataFrame

```python
emp_df.isnull().sum()
```

### i) Rename the Employee_ID column to ID

```python
emp_df.rename(columns={"Employee_ID": "ID"}, inplace=True)
emp_df
```

### j) Find all employees who:\n i. Have more than 5 years of experience\n ii. Belong to the IT department

```python
more_than_5_years = emp_df[emp_df["Years_of_Experience"] > 5]
print("i. Employees with more than 5 years of experience:")
display(more_than_5_years)

it_dept = emp_df[emp_df["Department"] == "IT"]
print("ii. Employees belonging to the IT department:")
display(it_dept)
```

### k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary

```python
emp_df["Tax"] = emp_df["Salary"] * 0.10
emp_df
```
